```python
import sys
sys.path.append("..")
from pipeline import extract, normalize, config

term = "حذف و اضافه"
for pdf_path in sorted(config.PDF_DIR.glob("*.pdf")):
    for p in extract.extract_pdf(pdf_path):
        text = normalize.clean_text(p["text"])
        if term in text:
            idx = text.find(term)
            print(f"FOUND in {pdf_path.name}, page {p['page']}")
            print("..." + text[max(0, idx-100):idx+200] + "...")
            print()
```
